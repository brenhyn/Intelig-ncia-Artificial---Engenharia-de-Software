19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o angulo entre eles (produto escalar normalizado).

```python
import numpy as np
GeradorNA1 = np.random.default_rng()

array191 = GeradorNA1.integers(1, 101, size=20)
array192 = GeradorNA1.integers(1, 101, size=20)

prodEsc = np.dot(array191, array192)
a = np.linalg.norm(array191)
b = np.linalg.norm(array192)

cosseno = prodEsc / (a * b)
radianos = np.arccos(cosseno)
anguloGraus = np.degrees(radianos)

print(anguloGraus)
```

1. Calcule a integral da função f (x) = x2 sin(x) no intervalo de 0 a π.

```python
import numpy as np
import scipy.integrate as integrate

def f(x):
  return x**2 *np.sin(x)

resultado, erro = integrate.quad(f, 0, np.pi)

print(f"Resultado: {resultado}\nErro aproximado: {erro}")
```

2. Resolva a equação diferencial dy
dx = x - y, com y(0) = 1, no intervalo de 0
a 10.

```python
import numpy as np
import scipy.integrate as integrate

def dyDx(x, y):
  return x - y

resultado = integrate.solve_ivp(dyDx, [0, 10], [1], t_eval=np.linspace(0, 10, 11))

valoresX = resultado.t
valoresY = resultado.y[0]

print(f"X: {valoresX}\nY: {valoresY}")
```

3. Encontre as raízes da função f (x) = x4 - 3x3 + 2 utilizando SciPy

```python
import numpy as np
import scipy.linalg as linalg

def f(x):
    return x**4 - 3*x**3 + 2

coe = [1, -3, 0, 0, 2]
matriz = np.polynomial.polynomial.polycompanion(coe[::-1])
raizes = linalg.eigvals(matriz)

print(raizes)
```

4. Use a função de otimização de SciPy para maximizar f (x) = -x2 + 4x + 1.

```python
import numpy as np
import scipy.optimize as optimize

def f(x):
  return -x**2 + x*4 + 1

def negativo(x):
  return -f(x)

r = optimize.minimize(negativo, 0)

maxX = r.x[0]
maxF = f(maxX)

print(f"X maximiza f: {maxX}\nMaximo de f: {maxF}")
```

5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
cos(4πt).

```python
import numpy as np
from scipy.fft import fft, fftfreq

pts = 10
t = np.arange(pts) / pts
sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)

y = fft(sinal)
freque = fftfreq(len(t), 1/pts)

pico = np.abs(y) > 1
print(f"Frequencias: {freque[pico]}\nAmplitudes: {np.abs(y[pico])}")
```

```python
import pandas as pd

prods ={
  'produto': ["Maçã", "Caramelo", "Uva", "Pão", "Ovo", "Vinho"],
  'Preço' : [5.99, 10.00, 5.99, 1.00, 0.50, 120.00],
  'Qtd' : [1, 2, 5, 3, 10, 1]
}

listaComp = pd.DataFrame(prods)
listaComp['Total'] = listaComp['Preço'] * listaComp['Qtd']
listaComp['Imposto'] = listaComp['Preço'] * 0.05

print(listaComp[listaComp['produto'].str.startswith('B')])
```

19. Crie um DataFrame a partir de um dicionário de listas e calcule a média de cada coluna numérica.

```python
import pandas as pd

dados = {
  'Filme': ['O labirinto do Fauno', 'Moonlight', 'Eyes Wide Shut', 'Brokeback Mountain'],
  'Duracao': [118, 111, 159, 134],
  'Nota': [8.2, 7.4, 7.5, 7.7]
}

imdb = pd.DataFrame(dados)
medias = imdb.select_dtypes(include="number").mean()

print(medias)
```

20. Agrupe os dados por uma coluna categórica e calcule a mediana de uma
coluna numérica

```python
import pandas as pd

dados = {
  'Filme': ['O labirinto do Fauno', 'Moonlight', 'Eyes Wide Shut', 'Brokeback Mountain'],
  'Categoria' : ['fantasia', 'Drama', 'Drama', 'Drama'],
  'Duracao': [118, 111, 159, 134],
  'Nota': [8.2, 7.4, 7.5, 7.7]
}

imdb = pd.DataFrame(dados)
medianaCategoria = (
  imdb.groupby("Categoria")["Duracao"].median().rename("medianaDuracao")
)

print(medianaCategoria)
```

8. Plote um gráfico de linha com intervalos de confiança (sombramento) para
a função y = sin(x)

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.fill_between(x, y - 0.2, y + 0.2, alpha=0.2)

plt.show()
```

18. Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

```python
import matplotlib.pyplot as plt
import numpy as np

categorias = ['A', 'B', 'C', 'D']
valores = np.array([-8, 10, 6, -5])
cores = ['g' if v >= 0 else 'r' for v in valores]

plt.bar(categorias, valores, color=cores)
plt.axhline(0, color='k', linewidth=0.8)
plt.show()
```

20. Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x2.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)
y = x**2

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xticks(np.arange(-5, 6, 1))
ax.set_yticks(np.arange(0, 26, 5))
ax.grid(True)

plt.show()
```
