21. Busca Binária


*    Implemente a busca binária para encontrar um elemento em uma
lista ordenada



```python
lista = [1,2,3,4,5,6,8]
numero = int(input("Digite um inteiro e descubra se ele esta na lista: "))
indexL = 0
indexR = len(lista) - 1
resposta = "Numero não esta na lista"

while indexL <= indexR:
    if lista[indexL] == numero:
      resposta = f"O numero foi encontrado no indice: {str(indexL)}"
      break
    else:
      if lista[indexL] < numero:
        indexL = indexL + 1

print(resposta)
```

Enquato a variavel indexL, inicialmente definida como 0, for menor ou igual a variavel indexR, definida por len(lista) -1, ou seja, o ultimo indice da lista, um loop executa o codigo que, caso o numero na lista com indice correspondente a indexL seja igual ao elemento procurado entao a variavel resposta recebe um novo valor com seu indice e o loop é interrompido, caso seja menor então indexL recebe o valor de si mesmo mais um. Após a execuçao do loop print imprime o valor da variável resposta.

22. Criptografia de Cifra de César


*   Crie uma função para criptografar uma mensagem usando a cifra de
César.



```python
def cifraDeCesar(texto, desloc):
  msg = ""
  for caractere in texto:
    if "a" <= caractere <= "z":
      indice = (ord(caractere)- ord('a') + desloc) % 26 + ord('a')
      msg = msg + (chr(indice))
    else:
      if "A" <= caractere <= "Z":
        indice = (ord(caractere)- ord('A') + desloc) % 26 + ord('A')
        msg = msg + (chr(indice))
      else:
        msg = msg + (caractere)
  return msg

pTexto = input("Digite o texto para criptografar: ")
pDesloc =int(input("Digite o valor para deslocamento: "))
print("Sua mensagem criptografada: ", cifraDeCesar(pTexto, pDesloc))
```

A funçao recebe como parametros, uma string e um int. Uma variável string msg é declarada vazia, e conforme um loop percorre a string, caso o caractere de cada execução do codigo seja uma letra, minuscula ou maiuscula, ela tranformada em um numero, como se fosse um indice do alfabeto, atraves de ord(), e a operação faz com que os indices vao de 0 ao 25, e então o indice e somado ao numero do deslocamento. o resultado % 26 é somado ao numero da letra "a", para caso o deslocamento tenha que percorrer o alfabeto mais de uma vez. Se o caractere nao for nem maiusculo, nem minusculo, ele não é uma letra e nao sofre alteração. Ao final de cada if ou else o resultado final do caractere em questao é tranformado novamente em caractere chr() e concatenado a variável msg que é retornada ao final da função.

24. Validação de CPF


*   Implemente uma função para validar um CPF brasileiro.



```python
def validarCpf(cpf):
  validade = "valido"
  if cpf[0] == cpf[1] == cpf[2] == cpf[3] == cpf[4] == cpf[5] == cpf[6] == cpf[7] == cpf[8] == cpf[9] == cpf[10]:
    validade = "invalido"
  else:

    dig1 = int(cpf[0])
    dig2 = int(cpf[1])
    dig3 = int(cpf[2])
    dig4 = int(cpf[3])
    dig5 = int(cpf[4])
    dig6 = int(cpf[5])
    dig7 = int(cpf[6])
    dig8 = int(cpf[7])
    dig9 = int(cpf[8])
    dig10 = int(cpf[9])
    dig11 = int(cpf[10])

    somaPriDig = (dig1 * 10 + dig2 * 9 + dig3 * 8 + dig4 * 7 + dig5 * 6 + dig6 * 5 + dig7 * 4 + dig8 * 3 + dig9 * 2)
    restoPriDig = (somaPriDig * 10) % 11
    if restoPriDig == 10:
      restoPriDig = 0

    somaSegDig = (dig1 * 11 + dig2 * 10 + dig3 * 9 + dig4 * 8 + dig5 * 7 + dig6 * 6 + dig7 * 5 + dig8 * 4 + dig9 * 3 + dig10 * 2)

    restoSegDig = (somaSegDig * 10) % 11
    if restoSegDig == 10:
        restoSegDig = 0

    if restoPriDig == dig10 and restoSegDig == dig11:
      validade = "valido"
    else:
      validade = "invalido"

  return(validade)

pCpf = input("Digite os 11 numeros do cpf em ordem e apenas os numeros: ")
print("O cpf é", validarCpf(pCpf))

```

Caso sejam iguais a variável validade é definida como "invalido", caso nao a funçao faz dois calculos para validar o cpf, o primeiro multiplica os nove primeiros numeros de 10 a 2, soma tudo, multiplica por 10 e tira o resto por 11 e ajustando para 0 caso o resto seja 10, o segundo faz o mesmo, porem, de 11 a 2 e incluindo o primeiro dos digitos verificadores, o digito 10, ja calculado na soma. Entao a funçao compara os dois valores calculados com os dois ultimos numeros do CPF informado, se as duas comparações forem verdadeiras, a funçao declara a variável validade como "valido", se nao, como "invalido"

*   Classifique uma lista de números em três categorias: positivos,

```python
def bubbleSort(listaNumeros):
  num = len(listaNumeros)
  for i in range(num):
    for j in range(num -1):
      if listaNumeros[j] > listaNumeros[j +1]:
        listaNumeros[j], listaNumeros[j + 1] = listaNumeros[j + 1], listaNumeros[j]
  return listaNumeros

lst = [1,2,5,4,7,6]
print(bubbleSort(lst))
```

A função recebe uma lista, uma variável num é definida pelo numero de elementos na lista. Um loop for é usado para executar, pela quantidade de elementos na lista, outro loop onde em cada execução o bloco de codigo compara o de um indice com o sucessivo, passando o vizinho maior para a proxima posição, ou seja, a cada execução do bloco de codigo do loop interno o maior vai para a ultima posição, com o loop externo executando esse loop para que todos os elementos da lista sejam comparados entre si. ao final a função retorna a lista.

37. Funções Recursivas


*   Implemente uma função recursiva para calcular o fatorial de um
número.



```python
def calcFatorial(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num * calcFatorial(num - 1)

print(calcFatorial(16))
```

 A função verifica se o numero é 0 ou 1 e retorna 1 nesse caso. Se esse nao for o caso, ela retorna o numero vezes fatorial do numero menos um, chamando si mesma, essas camadas vao se acumulando até o numero chegar em 1 e todos os resultado começam a ser multiplicados de tras pra frente devolvendo o fatorial do numero inicial.

48. Ordenação por Inserção


*   Implemente o algoritmo de ordenação por inserção (Insertion Sort).



```python
def ordInserc(listaOrdenar):
  for i in range(1, len(listaOrdenar)):
    segundoElem = listaOrdenar[i]
    primeiroElem = i - 1
    while primeiroElem >= 0 and listaOrdenar[primeiroElem] > segundoElem:
      listaOrdenar[primeiroElem + 1] = listaOrdenar[primeiroElem]
      primeiroElem = primeiroElem - 1
    listaOrdenar[primeiroElem + 1] = segundoElem
  return listaOrdenar

listaord = [1,2,3,4,5,0,8,7,6]

print(ordInserc(listaord))
```

Um for percorre a lista comparando os elementos a partir do segundo, com seu antecessor, enquanto o indice do antecessor for maior ou igual a zero, variavel primeiroElem, e o elemento for maior que o elemento que o sucede, o sucessor troca de lugar com o antecessor. Ao final de todas as comparações a lista é retornada.
